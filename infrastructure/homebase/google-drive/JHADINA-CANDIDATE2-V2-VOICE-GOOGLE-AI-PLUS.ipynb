## 4. Generate 02A–02D privately with per-file SHA-256 receipts

This cell runs actual GPU synthesis only after manual compute approval; it saves four new WAVs, four MP3s and a growing manifest in a NEW folder beneath the existing private archive. If a run fails, partial files stay quarantined; they are never published.

```python
budget_gate()
assert 'tts' in globals(), "Run the Qwen model loader first"
import numpy as np,random,soundfile as sf,datetime,hashlib,subprocess,json
from IPython.display import Audio,display
OUT=ARCHIVE/('Candidate2-V2-Colab-'+datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%dT%H%M%SZ'))
OUT.mkdir(exist_ok=False)
manifest={'identity':'voice:jhadina:canonical:v1','profile':'jhadina:canonical',
 'packId':PACK['packId'],'sourceReference':PACK['reference']['candidateId'],
 'origin':'owner_google_ai_plus_colab_manual','state':'candidate_unapproved',
 'productionApproved':False,'speakerSimilarityMeasured':False,'qualityClaim':False,'candidates':[]}
for i,v in enumerate(PACK['variants'],start=1):
    seed=v['seed'];torch.manual_seed(seed);torch.cuda.manual_seed_all(seed)
    np.random.seed(seed);random.seed(seed)
    wavs,sr=tts.generate_voice_design(text=PACK['scripts']['common'],language='English',instruct=v['instruction'])
    wav=np.asarray(wavs[0])
    assert wav.size>0 and np.isfinite(wav).all(), "Model produced invalid samples"
    wavpath=OUT/f'candidate-{i}.wav';mp3path=OUT/f'candidate-{i}.mp3'
    sf.write(str(wavpath),wav,sr)
    subprocess.run(['ffmpeg','-nostdin','-hide_banner','-loglevel','error','-y','-i',str(wavpath),'-codec:a','libmp3lame','-q:a','3',str(mp3path)],check=True)
    raw=wavpath.read_bytes();mp3=mp3path.read_bytes()
    assert raw[:4]==b'RIFF' and raw[8:12]==b'WAVE' and len(mp3)>512 and (mp3[:3]==b'ID3' or (mp3[0]==255 and mp3[1]&224==224))
    wavhash=hashlib.sha256(raw).hexdigest();mp3hash=hashlib.sha256(mp3).hexdigest()
    manifest['candidates'].append({'id':f'jhadina-audition:{seed}','seed':seed,
       'variantId':v['id'],'name':v['name'],'file':wavpath.name,'mp3File':mp3path.name,
       'artifactSha256':wavhash,'mp3Sha256':mp3hash,'provider':'qwen3-tts',
       'modelId':PACK['provider'],'providerTaskId':f'colab-local:{seed}:{wavhash[:16]}',
       'voiceIdentityId':manifest['identity'],'state':'candidate_unapproved',
       'qualityClaim':False,'speakerSimilarityMeasured':False,'productionApproved':False,
       'durationSeconds':round(wav.size/sr,2)})
    (OUT/'manifest.json').write_text(json.dumps(manifest,indent=2,ensure_ascii=False)+'\n')
    print(v['name'],round(wav.size/sr,2),'s',mp3path.name);display(Audio(filename=str(mp3path)))
    torch.cuda.empty_cache()
assert len(manifest['candidates'])==4
print("UNAPPROVED AUDIO SET READY:",OUT)
print("Disconnect Colab GPU now; do not promote voice or publish without owner review.")
```
